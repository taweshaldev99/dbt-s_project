```python
import	pandas	as	pd,	numpy	as	np
rng	=	np.random.default_rng(42)
countries	=	["NP","US","IN","DE","JP","BR","GB","AU"]
#	----	customers.csv	(~10k	rows)	----
n_c	=	10_000
customers	=	pd.DataFrame({
				"customer_id":	np.arange(1,	n_c	+	1),
				"name":	[f"customer_{i}"	for	i	in	range(1,	n_c	+	1)],
				"country":	rng.choice(countries,	n_c),
				"signup_date":	(pd.Timestamp("2019-01-01")
								+	pd.to_timedelta(rng.integers(0,	2500,	n_c),	unit="D")).date,
})
#	----	orders.csv	(~300k	rows,	60%	skew	to	NP:	Day	3	is	baked	in	NOW)	----
n_o	=	300_000
p	=	[0.60]	+	[0.40	/	7]	*	7
orders	=	pd.DataFrame({
				"order_id":	np.arange(1,	n_o	+	1),
				"customer_id":	rng.integers(1,	n_c	+	1,	n_o),
				"product_id":	rng.integers(1,	500,	n_o),
				"amount":	np.round(rng.exponential(80,	n_o)	+	1,	2),
				"ts":	pd.Timestamp("2026-01-01")	+	pd.to_timedelta(
								rng.integers(0,	200	*	24	*	3600,	n_o),	unit="s"),
				"country":	rng.choice(countries,	n_o,	p=p),
})
path	=	"/Volumes/workspace/default/spark_course"
customers.to_csv(f"{path}/customers.csv",	index=False)
orders.to_csv(f"{path}/orders.csv",	index=False)
#	Parquet	copy
spark.createDataFrame(orders).write.mode("overwrite")	\
					.parquet(f"{path}/orders_parquet")
print("Seed	complete.	Verify	files	in	Catalog	→	workspace	→	default	→	spark_course")
```
